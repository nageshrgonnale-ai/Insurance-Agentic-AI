```python
import os
print(os.listdir("/content/drive/MyDrive"))
```

```python
def extract_text(pdf_path):

    doc=fitz.open(pdf_path)

    text=""

    for page in doc:

        text+=page.get_text()

    return text
```

```python
sample=os.path.join(PDF_FOLDER,pdfs[0])

text=extract_text(sample)

print(text[:3000])
```

```python
documents=[]

for pdf in pdfs:

    path=os.path.join(PDF_FOLDER,pdf)

    txt=extract_text(path)

    documents.append(
        {
            "filename":pdf,
            "text":txt
        }
    )

print(len(documents))
```

```python
import pandas as pd
df=pd.DataFrame(documents)

df.head()
```

```python
df.to_csv(
    "/content/drive/MyDrive/Insurance-Agentic-AI/data/extracted_text.csv",
    index=False
)

print("Saved Successfully")
```
